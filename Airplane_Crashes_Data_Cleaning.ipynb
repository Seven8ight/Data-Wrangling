4. Each column except the first one has some missing data. For each column, name
the most appropriate method for treating the missing data. Justify your choices.

    For most columns, some rows are empty hence deducing the missing mechanism is Missing at Random.

    For column Time: We can perform a constant imputation or Binning such that we fill the row element with a predefined fixed value since a placeholder preserves the row for other analyses.

    For column Flight: Most rows have no value so we can delete the column instead of imputation since old flights, as per the data, did not have any flight numbers, imputattion would be useless.

    For column Route: We can impute it as well since the placeholder retains the row's target variables without faking any geographical data.

    For column Registration: We can delete the column or impute it with placeholders. Since the data is missing randomly and the values are unique alphanumeric identifiers that can't be statistically modelled, dropping the clumn may prevent false data provision whilst acknowledging historical gaps.

    For columns Location, Operator and Type: They experience the same missing data cause i.e. Missing not at random and can be treated via Listwise Deletion such that dropping the rows for the following columns causes manageable data loss. Imputing cannot help since a column like location cannot be statistically guessed on where an unrecovered plane is and for operator and type imputing placeholders would assign historical crashes to the wrong airlines, rendering the dataset unreliable and reducing its integrity.

    For column Aboard: We can use grouped median imputation in that the data is missing at random hence you can use the Type column to find the median passenger count for that aircraft providing an accurate estimation without losing any row.

    For column Fatalities: We can use Listwise deletion. Fatalities is a target variable for most analyses of the dataset. Imputing data here can cause significant change in the models perceived accuracy thus dropping rows is safer and more convenient for this column.

    For column Summary: We can use imputation since its an unstructued free-form data element. Providing it a standard text placeholdder allows algorithm to process the data without discarding the valuable numerical data in those rows.

8. In the ‘fatality_locations’ dataframe, the location column has data items
represented as ‘Region, U.S. State/Country’. Split this column to have two columns
of regions and states/countries.

```python
import pandas as pd
import numpy as np

# Build the fatality_locations dataframe
fatality_locations = df[['Date', 'Location', 'Aboard', 'Fatalities']].copy()

# Split the Location column from the first comma
split_cols = fatality_locations['Location'].str.split(',', n=1, expand=True)

fatality_locations['Region'] = split_cols[0].str.strip()
fatality_locations['State_Country'] = split_cols[1].str.strip()

print(fatality_locations.head(10))
```

9. Order the dataframe by the number of fatalities from highest to lowest and select the
first 100.

```python
import pandas as pd

top_100_fatalities = fatality_locations.sort_values(by='Fatalities', ascending=False).head(100)

print(top_100_fatalities.head(10))  # just previewing the first 10 of the 100
print(f"Shape of the new dataframe: {top_100_fatalities.shape}")
```

10. Generate a pie chart that shows the distribution of the top 25 number of fatalities
per country/U.S. State. [Hint: charts can be generated from the seaborn library]

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Take the top 25
top_25 = top_100_fatalities.head(25)

# Use State_Country as the label if missing, use Region
labels = top_25['State_Country'].fillna(top_25['Region'])

plt.figure(figsize=(9, 9))
colors = sns.color_palette('pastel', len(top_25))
plt.pie(top_25['Fatalities'], labels=labels, autopct='%1.1f%%', colors=colors)
plt.title('Distribution of Top 25 Fatalities by Country/U.S. State')
plt.tight_layout()
plt.show()
```
